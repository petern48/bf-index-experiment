```python
%matplotlib inline
import json
import os

from utils import (
    DEFAULT_DATA,
    DEFAULT_OUT,
    filter_data_by_size,
    plot_all_grid,
)
```

```python
display_inline = True  # Set to False to save to disk instead
input_dir = DEFAULT_DATA
output_dir = DEFAULT_OUT


with open(input_dir) as f:
    data = json.load(f)

os.makedirs(output_dir, exist_ok=True)
```

```python
if display_inline:
    print("Displaying graphs inline")
else:
    print(f"Generating graphs in {output_dir}/")

# Generate separate figure sets for each dataset size (small, large)
for size in data["dataset_sizes"]:
    print(f"\n--- {size} dataset ---")
    filtered = filter_data_by_size(data, size)
    # Grid layout: row1=read_row_groups, read_datafiles | row2=time_write, time_read
    #              row3=memory_write, memory_read | row4=disk_storage (full width)
    plot_all_grid(filtered, output_dir, display_inline=display_inline, dataset_size=size)
```
