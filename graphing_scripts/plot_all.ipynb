```python
%matplotlib inline
import json
import os

from utils import (
    DEFAULT_DATA,
    DEFAULT_OUT,
    _experiment_list,
    filter_data_by_experiment,
    filter_data_by_size,
    plot_all_grid,
)
```

```python
display_inline = True  # Set to False to save to disk instead
input_dir = DEFAULT_DATA
output_dir = DEFAULT_OUT


with open(input_dir) as f:
    data = json.load(f)

os.makedirs(output_dir, exist_ok=True)
```

```python
if display_inline:
    print("Displaying graphs inline")
else:
    print(f"Generating graphs in {output_dir}/")

# Generate separate figure for each experiment
experiments = _experiment_list(data)
for exp_id in experiments:
    print(f"\n--- {exp_id} ---")
    filtered = filter_data_by_experiment(data, exp_id)
    plot_all_grid(filtered, output_dir, display_inline=display_inline, dataset_size=exp_id)
```
